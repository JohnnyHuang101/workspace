```python
import os
import uuid
import queue
import threading
import time
from collections import defaultdict
from concurrent.futures import ThreadPoolExecutor

import numpy as np
import psutil
import pypdf
import spacy
from qdrant_client import QdrantClient
from qdrant_client.models import PointStruct


# ── Tunable constants ──────────────────────────────────────────────────────────

EMBED_BATCH_SIZE  = 128     # sentences fed to GPU at once — tune to VRAM
UPSERT_BATCH_SIZE = 32     # chunks per Qdrant request
NLP_WORKERS       = 4      # parallel pypdf+spaCy threads (raise to os.cpu_count())
CHUNK_WORKERS     = 2      # parallel cosine-sim threads (numpy releases GIL — scale freely)
COSINE_THRESHOLD  = 0.65    # semantic distance above which a split is inserted
COLLECTION_NAME   = "ai_knowledge"
MONITOR_INTERVAL  = 15.0    # seconds between monitor printouts
MAX_TOKENS = 512   # add near your other constants


# ─────────────────────────────────────────────────────────────────────────────
# Stage 0 — Monitor  (background thread — no queue, reads all three)
# ─────────────────────────────────────────────────────────────────────────────

def _monitor_worker(
    sent_q: queue.Queue,
    embed_out_q: queue.Queue,
    upsert_q: queue.Queue,
    stop_event: threading.Event,
    interval: float = MONITOR_INTERVAL,
) -> None:
    """
    Polls every `interval` seconds and prints a one-line dashboard:
      queue depths | GPU VRAM alloc/reserved/total | system RAM

    Reading queue depths:
      - sent_q   near max → NLP is fast, embedder is the bottleneck
      - embed_q  near max → embedder is fast, chunking workers are the bottleneck
      - upsert_q near max → chunking is fast, Qdrant writes are the bottleneck
      - all queues empty  → everything balanced (or pipeline stalled — watch RAM)
    """
    proc = psutil.Process(os.getpid())
    has_cuda = torch.cuda.is_available()

    while not stop_event.is_set():
        sq = sent_q.qsize()
        eq = embed_out_q.qsize()
        uq = upsert_q.qsize()

        if has_cuda:
            alloc    = torch.cuda.memory_allocated()  / 1024**3
            reserved = torch.cuda.memory_reserved()   / 1024**3
            total    = torch.cuda.get_device_properties(0).total_memory / 1024**3
            vram_str = f"VRAM {alloc:.2f}/{reserved:.2f}/{total:.1f} GB (alloc/reserv/total)"
        else:
            vram_str = "VRAM N/A (no CUDA)"

        ram_gb = proc.memory_info().rss / 1024**3

        print(
            f"[MON] "
            f"sent_q={sq:>4}/{sent_q.maxsize}  "
            f"embed_q={eq:>3}/{embed_out_q.maxsize}  "
            f"upsert_q={uq:>4}/{upsert_q.maxsize} | "
            f"{vram_str} | "
            f"RAM {ram_gb:.2f} GB"
        )
        stop_event.wait(interval)


# ─────────────────────────────────────────────────────────────────────────────
# Stage 1 — NLP Workers  (thread pool, CPU-bound / GIL released by spaCy C exts)
# ─────────────────────────────────────────────────────────────────────────────



# ─────────────────────────────────────────────────────────────────────────────
# Stage 2 — Embedder  (single thread — owns the GPU)
# ─────────────────────────────────────────────────────────────────────────────

def _embedder_worker(
    embedder,
    sent_q: queue.Queue,
    embed_out_q: queue.Queue,
    n_chunk_workers: int,
    error_event: threading.Event,
) -> None:
    has_cuda = torch.cuda.is_available()
    pending: list[dict] = []

    def _flush(items: list[dict]) -> None:
        texts = [it["text"] for it in items]
        with torch.inference_mode():
            vecs = embedder.encode(
                texts,
                batch_size=EMBED_BATCH_SIZE,
                show_progress_bar=False,
                convert_to_numpy=True,
            )
        if has_cuda:
            torch.cuda.empty_cache()

        groups: dict[str, list] = defaultdict(list)
        for item, vec in zip(items, vecs):
            groups[item["doc_id"]].append({**item, "vector": vec})
        for doc_items in groups.values():
            doc_items.sort(key=lambda x: x["pos"])
            embed_out_q.put(doc_items)

    try:
        while True:
            item = sent_q.get()
            if item is None:
                if pending:
                    _flush(pending)
                break
            pending.append(item)
            if len(pending) >= EMBED_BATCH_SIZE:
                _flush(pending)
                pending = []
    except Exception as e:
        print(f"[Embedder] CRASHED: {type(e).__name__}: {e}")
        error_event.set()
    finally:
        # guaranteed to run even on crash — chunkers will never deadlock
        for _ in range(n_chunk_workers):
            embed_out_q.put(None)
        print("[Embedder] Done")

# ─────────────────────────────────────────────────────────────────────────────
# Stage 3 — Chunking Workers  (thread pool, CPU-bound / numpy releases GIL)
# ─────────────────────────────────────────────────────────────────────────────

def _chunking_worker(embed_out_q: queue.Queue, upsert_q: queue.Queue) -> None:
    """
    Receive an ordered list of (sentence, vector) pairs for one doc-batch.
    Compute pairwise cosine distances; insert splits where distance > threshold.

    FIX: no longer re-broadcasts the None sentinel. The embedder now puts exactly
    CHUNK_WORKERS sentinels so each worker receives exactly one and exits cleanly.
    """
    while True:
        doc_items = embed_out_q.get()
        if doc_items is None:
            break                           # this worker's sentinel — exit cleanly
        _split_and_enqueue(doc_items, upsert_q)


def _split_and_enqueue(doc_items: list[dict], upsert_q: queue.Queue) -> None:
    vecs = np.array([it["vector"] for it in doc_items], dtype=np.float32)

    if len(doc_items) == 1:
        _push_chunk(doc_items, vecs, 0, 1, upsert_q)
        return

    # Normalise for cosine arithmetic
    norms = np.linalg.norm(vecs, axis=1, keepdims=True)
    norms[norms == 0] = 1.0
    norm_vecs = vecs / norms

    # Pairwise cosine distance between adjacent sentences
    cosines   = np.sum(norm_vecs[:-1] * norm_vecs[1:], axis=1)
    distances = 1.0 - cosines
    splits    = np.where(distances > COSINE_THRESHOLD)[0]

    # Build chunk boundaries [0, split+1, ..., N]
    boundaries = [0, *(int(s) + 1 for s in splits), len(doc_items)]
    for i in range(len(boundaries) - 1):
        _push_chunk(doc_items, vecs, boundaries[i], boundaries[i + 1], upsert_q)


def _push_chunk(
    doc_items: list[dict],
    vecs: np.ndarray,
    start: int,
    end: int,
    upsert_q: queue.Queue,
) -> None:
    segment    = doc_items[start:end]
    chunk_text = " ".join(it["text"] for it in segment)

    # Mean-pool sentence vectors → normalise → chunk embedding
    chunk_vec = np.mean(vecs[start:end], axis=0)
    norm      = np.linalg.norm(chunk_vec)
    if norm > 0:
        chunk_vec /= norm

    upsert_q.put(PointStruct(
        id=str(uuid.uuid4()),
        vector=chunk_vec.tolist(),
        payload={
            "text":        chunk_text,
            "source_file": segment[0]["doc_id"],
            "page_n":      segment[0]["page"],
        },
    ))


# ─────────────────────────────────────────────────────────────────────────────
# Stage 4 — Qdrant Upsert Worker  (single thread, network I/O)
# ─────────────────────────────────────────────────────────────────────────────

def _upsert_worker(vector_db: QdrantClient, upsert_q: queue.Queue) -> None:
    """
    Drain upsert_q and write to Qdrant in UPSERT_BATCH_SIZE batches.
    Runs on its own thread so chunking workers are never stalled on network I/O.
    """
    batch: list[PointStruct] = []
    total = 0

    def _flush() -> None:
        nonlocal total
        vector_db.upsert(collection_name=COLLECTION_NAME, wait=True, points=batch)
        total += len(batch)
        # print(f"[Qdrant] {len(batch)} chunks upserted (running total: {total})")
        batch.clear()

    while True:
        item = upsert_q.get()
        if item is None:
            if batch:
                _flush()
            break
        batch.append(item)
        if len(batch) >= UPSERT_BATCH_SIZE:
            _flush()

    print(f"[Qdrant] Done. Total chunks: {total}")


def get_pdf_tasks(pdf_paths: list[str]):
    """Yields (file_path, page_num) pairs; pre-reads page count once per file."""
    for file_path in pdf_paths:
        try:
            with open(file_path, "rb") as fh:
                reader = pypdf.PdfReader(fh, strict=False)  # tolerate xref errors
                num_pages = len(reader.pages)
            for page_num in range(num_pages):
                yield file_path, page_num
        except Exception as e:
            print(f"[Orchestrator] Skipping {file_path}: {e}")

def _nlp_worker(
    nlp,
    sent_q: queue.Queue,
    error_event: threading.Event,
    file_path: str,
    page_num: int,
) -> None:
    if error_event.is_set():
        return

    filename = os.path.basename(file_path)
    try:
        with open(file_path, "rb") as fh:
            reader = pypdf.PdfReader(fh, strict=False)  # same: tolerate xref errors
            raw = reader.pages[page_num].extract_text() or ""

        if not raw.strip():
            # Log scanned/blank pages so you can tell if ALL pages are empty
            print(f"[NLP] No text on {filename} p{page_num + 1} (scanned or blank)")
            return

        pos = 0
        for sent in nlp(raw).sents:
            text = sent.text.strip()
            if not text:
                continue
            tokens = text.split()
            chunks = (
                [" ".join(tokens[i : i + MAX_TOKENS]) for i in range(0, len(tokens), MAX_TOKENS)]
                if len(tokens) > MAX_TOKENS
                else [text]
            )
            for chunk in chunks:
                sent_q.put({"doc_id": filename, "page": page_num + 1, "pos": pos, "text": chunk})
                pos += 1

    except Exception as e:
        # Re-raise so fut.result() surfaces it instead of hiding it
        raise RuntimeError(f"[NLP] {filename} p{page_num}: {e}") from e
    
def ingest_pdfs(
    pdf_paths: list[str],
    vector_db: QdrantClient,
    embedder,       # any sentence-transformers–compatible encoder
    nlp,            # spaCy Language object
) -> None:
    """
    Ingest a list of PDF files into Qdrant using a parallel fan-out/fan-in pipeline.

    Thread layout:
      1 monitor thread      — prints queue depths + memory every MONITOR_INTERVAL s
      NLP_WORKERS threads   — pypdf + spaCy  (CPU, GIL released by spaCy C extensions)
      1 embedder thread     — GPU encode      (single owner of the model)
      CHUNK_WORKERS threads — cosine sim      (CPU, GIL released by numpy)
      1 upsert thread       — Qdrant network  (I/O, never blocks GPU or CPU workers)

    Sentinel propagation (fixed — no re-broadcast race):
      NLP done          → sent_q.put(None)           (one pill)
      Embedder sees it  → embed_out_q.put(None) × CHUNK_WORKERS  (one per chunker)
      Each chunker gets its own None and exits
      After all chunkers join → upsert_q.put(None)   (one pill)
      Upsert worker exits
    """
    # Bounded queues provide automatic backpressure between stages.
    # If a downstream stage is slow, its input queue fills and the upstream
    # producer blocks — preventing unbounded RAM growth.
    sent_q      = queue.Queue(maxsize=768)   # reduced from 400 to limit peak RAM
    embed_out_q = queue.Queue(maxsize=20)
    upsert_q    = queue.Queue(maxsize=500)
    error_event = threading.Event()


    # ── Monitor (start first so it captures the whole run) ───────────────────
    stop_monitor = threading.Event()
    monitor_thread = threading.Thread(
        target=_monitor_worker,
        args=(sent_q, embed_out_q, upsert_q, stop_monitor),
        name="monitor",
        daemon=True,
    )
    monitor_thread.start()

    # ── Start downstream workers first (they block on empty queues until fed) ─
    chunk_threads = [
        threading.Thread(
            target=_chunking_worker,
            args=(embed_out_q, upsert_q),
            name=f"chunker-{i}",
            daemon=True,
        )
        for i in range(CHUNK_WORKERS)
    ]

    upsert_thread = threading.Thread(
        target=_upsert_worker,
        args=(vector_db, upsert_q),
        name="qdrant",
        daemon=True,
    )
    embed_thread = threading.Thread(
        target=_embedder_worker,
        args=(embedder, sent_q, embed_out_q, CHUNK_WORKERS, error_event),
        name="embedder",
        daemon=True,
    )

    for t in chunk_threads:
        t.start()
    upsert_thread.start()
    embed_thread.start()


    # ── Run NLP workers; block until all PDFs are fully segmented ─────────────
    with ThreadPoolExecutor(max_workers=NLP_WORKERS, thread_name_prefix="nlp") as pool:
        futures = {
            pool.submit(_nlp_worker, nlp, sent_q, error_event, fp, pn): (fp, pn)
            for fp, pn in get_pdf_tasks(pdf_paths)
            if not error_event.is_set()
        }

        failed = 0
        for fut in futures:
            try:
                fut.result()
            except Exception as e:
                failed += 1
                print(repr(e))   # repr() gives you the full chain including the cause

        print(f"[Orchestrator] {failed} failed / {len(futures)} total tasks")
        print(f"[Orchestrator] sent_q final size: {sent_q.qsize()}")

    # ── Propagate sentinel down the pipeline ──────────────────────────────────
    sent_q.put(None)        # → embedder exits → puts CHUNK_WORKERS Nones in embed_out_q
    embed_thread.join()     # wait for embedder to finish and send its sentinels

    for t in chunk_threads:
        t.join()            # each chunker exits after receiving its own None

    upsert_q.put(None)      # → upsert worker flushes remaining batch and exits
    upsert_thread.join()

    # ── Stop monitor ──────────────────────────────────────────────────────────
    stop_monitor.set()
    monitor_thread.join()

    print("\n✓ Ingestion complete.")

```

```python
from qdrant_client.models import VectorParams, Distance
# Requires transformers>=4.51.0
# Requires sentence-transformers>=2.7.0

from sentence_transformers import SentenceTransformer

# # Load the model
model = SentenceTransformer("Qwen/Qwen3-Embedding-0.6B", device='cuda')

# 2. Initialize Models and Client
print("Initializing environment components...")
nlp_model = spacy.load("en_core_web_sm")
embedder_model = model
qdrant_client = QdrantClient(url="http://localhost:6333")

# 3. Setup Qdrant Collection Structure
print("Re-creating Qdrant collection 'ai_knowledge'...")
qdrant_client.recreate_collection(
    collection_name="ai_knowledge",
    vectors_config=VectorParams(size=1024, distance=Distance.COSINE)
)

file_paths = [os.path.join(os.getcwd(),"ai_pdfs",f) for f in os.listdir(os.path.join(os.getcwd(), "ai_pdfs"))]
```

```python
ingest_pdfs(file_paths,qdrant_client,embedder_model,nlp_model)
```
